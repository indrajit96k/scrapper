```python
# Import Libraries
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://webscraper.io/test-sites/e-commerce/static"

response = requests.get(url)
soup = BeautifulSoup(
    response.text,
    'html.parser'
)
products = soup.find_all(
    'div',
    class_='thumbnail'
)

product_list = []
for product in products:
    title = product.find('a', class_='title')
    if title:
        title = title.text.strip()
    else:
        title = "N/A"

    price = product.find('h4', class_='price')
    if price:
        price = price.text.strip()
    else:
        price = "N/A"

    description = product.find('p',class_='description')
    if description:
        description = description.text.strip()
    else:
        description = "N/A"

    ratings = product.find_all('span', class_='glyphicon-star')
    rating_count = len(ratings)

    review_data = {
        'Title': title,
        'Price': price,
        'Description': description,
        'Ratings': rating_count
    }
    product_list.append(review_data)

df = pd.DataFrame(product_list)
print(df.head())

df.to_csv(
    "ecommerce_reviews.csv",
    index=False
)
print("\nData Saved Successfully!")
```
