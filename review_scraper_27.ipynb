```python
import requests
import pandas as pd

url = "https://store.steampowered.com/appreviews/730?json=1"

response = requests.get(url)
data = response.json()
print(data["reviews"][0])
reviews = data['reviews']

review_list = []

for review in reviews:
    customer_name=review['author']['personaname']
    customer_id = review['author']['steamid']
    comment = review['review']
    votes_up = review['votes_up']
    rating = review['voted_up']

    review_data = {
        'customer_name':customer_name,
        'Customer Steam ID': customer_id,
        'Comment': comment,
        'Votes Up': votes_up,
        'Recommended': rating
    }

    review_list.append(review_data)

df = pd.DataFrame(review_list)

print(df.head(10))

df.to_csv("steam_reviews.csv", index=False)

print("\nSteam Reviews Saved Successfully!")
```
