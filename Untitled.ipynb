```python
import pandas as pd
import requests
url="https://store.steampowered.com/appreviews/730?json=1"
response=requests.get(url)
data=response.json()
reviews=data["reviews"]
print(reviews)
review_list=[]
for review in reviews:
    username=review["author"]["personaname"]
    userid=review["author"]["steamid"]
    comment=review["review"]
    voted_up=review["votes_up"]
    rating=review["voted_up"]
    review_data={
        "username":username,
        "userid":userid,
        "comment":comment,
        "voted_up":voted_up,
        "rating":rating
        
    }
    review_list.append(review_data)
df=pd.DataFrame(review_list)
df.to_csv("review_list.csv") 
print("done")
    
```
